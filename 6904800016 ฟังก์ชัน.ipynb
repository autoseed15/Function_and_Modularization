```python
#ข้อ1
def is_valid_score(score: float) -> bool:

    return 0 <= score <= 100


def to_grade(score: float) -> str:

    if not is_valid_score(score):
        return "คะแนนไม่ถูกต้อง"
    if score >= 80:
        return "A"
    elif score >= 70:
        return "B"
    elif score >= 60:
        return "C"
    return "F"


def print_result(student_id: str, score: float) -> None:

    print(f"รหัส: {student_id} | คะแนน: {score} | เกรด: {to_grade(score)}")


# ตัวอย่างการใช้งาน
if __name__ == "__main__":
    print_result("6501001", 60)
    print_result("6501002", 105)
```

```python
#ข้อ2
def safe_average(total, count):
    if count <= 0:
        return None
    return round(total / count, 2)

d1 = float(input("วันที่ 1: "))
d2 = float(input("วันที่ 2: "))
d3 = float(input("วันที่ 3: "))
d4 = float(input("วันที่ 4: "))
d5 = float(input("วันที่ 5: "))

total = 0
count = 0

if d1 >= 0:
    total += d1
    count += 1
if d2 >= 0:
    total += d2
    count += 1
if d3 >= 0:
    total += d3
    count += 1
if d4 >= 0:
    total += d4
    count += 1
if d5 >= 0:
    total += d5
    count += 1

avg = safe_average(total, count)

if avg is None:
    print("ไม่มีข้อมูลที่ใช้ได้")
else:
    print("ค่าเฉลี่ย PM2.5 =", avg)
```
